```python
#benchmark against common methods for collaborative filtering
#three methods will be used
#matrix factorization(svd)
#label-propogation(semi supervised)
#highbrid model specifiaclly light-fm(hybrid matrix facorixzation using NN)
import os
datafolder = "datafiles"
final_df_path = os.path.join(datafolder,"final_df.pkl")
#check if the file exists
if os.path.exists(final_df_path):
    #load the final dataframe
    final_df = pd.read_pickle(final_df_path)
```

```python
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity

# Compute affinity matrix
user_item_matrix = final_df.pivot_table(index="user", columns="item", values="rating").fillna(0)
W = cosine_similarity(user_item_matrix)
np.fill_diagonal(W, 0)  # No self-similarity

# Label propagation
D = np.diag(W.sum(axis=1) + 1e-5)  # Degree matrix
Y = user_item_matrix.values.copy()  # Known ratings
Y_unknown = np.isnan(Y)

for _ in range(20):  # Iterate until convergence
    Y_new = np.linalg.solve(D, W @ Y)
    Y[Y_unknown] = Y_new[Y_unknown]
```
