```python
import pandas as pd
#benchmark against common methods for collaborative filtering
#three methods will be used
#matrix factorization(svd)
#label-propogation(semi supervised)
#highbrid model specifiaclly light-fm(hybrid matrix facorixzation using NN)
import os
datafolder = "datafiles"
final_df_path = os.path.join(datafolder,"final_df.pkl")
#check if the file exists
if os.path.exists(final_df_path):
    #load the final dataframe
    df = pd.read_pickle(final_df_path)
```

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import GroupShuffleSplit
from sklearn.metrics import mean_squared_error, mean_absolute_error



# Split items per user into train/test
splitter = GroupShuffleSplit(n_splits=1, test_size=0.2, random_state=42)
train_idx, test_idx = next(splitter.split(df, groups=df["user"]))

train_df = df.iloc[train_idx]
test_df = df.iloc[test_idx]

# Create masked user-item matrices (train has NaNs for test items)
user_item_train = train_df.pivot(index="user", columns="item", values="rating")
user_item_test = test_df.pivot(index="user", columns="item", values="rating")

# Align matrices to include all users and items
all_users = df["user"].unique()
all_items = df["item"].unique()
user_item_train = user_item_train.reindex(index=all_users, columns=all_items)
user_item_test = user_item_test.reindex(index=all_users, columns=all_items)

# Convert to numpy arrays (NaN for unobserved entries)
R_train = user_item_train.values
R_test = user_item_test.values
```

```python
from sklearn.decomposition import TruncatedSVD

# Fill NaN with 0 (or user/item mean)
R_train_filled = np.nan_to_num(R_train, nan=0)

# Fit SVD on training data
n_components = 50
svd = TruncatedSVD(n_components=n_components)
user_factors = svd.fit_transform(R_train_filled)
item_factors = svd.components_

# Predict ratings: U * V^T
R_pred_svd = user_factors @ item_factors

# Extract test predictions and true ratings
test_mask = ~np.isnan(R_test)
y_true = R_test[test_mask]
y_pred_svd = R_pred_svd[test_mask]

# Calculate metrics
rmse_svd = np.sqrt(mean_squared_error(y_true, y_pred_svd))
mae_svd = mean_absolute_error(y_true, y_pred_svd)
```

```python
from sklearn.metrics.pairwise import cosine_similarity

# Compute user similarity matrix on training data
W_train = cosine_similarity(R_train_filled)

# Predict test ratings using similar users' ratings
R_pred_lp = np.zeros_like(R_test)
for (u, i) in zip(*np.where(test_mask)):
    # Find users who rated item i in the train set
    rated_users = np.where(~np.isnan(R_train[:, i]))[0]
    if len(rated_users) == 0:
        # Fallback to global mean if no ratings
        R_pred_lp[u, i] = np.nanmean(R_train)
    else:
        # Weighted average of similar users' ratings
        similarities = W_train[u, rated_users]
        ratings = R_train[rated_users, i]
        R_pred_lp[u, i] = np.dot(similarities, ratings) / (np.sum(similarities) + 1e-9)

# Extract predictions and calculate metrics
y_pred_lp = R_pred_lp[test_mask]
rmse_lp = np.sqrt(mean_squared_error(y_true, y_pred_lp))
mae_lp = mean_absolute_error(y_true, y_pred_lp)
```

```python
results = {
    "Method": ["SVD", "Label Propagation"],
    "RMSE": [rmse_svd, rmse_lp],
    "MAE": [mae_svd, mae_lp]
}

results_df = pd.DataFrame(results)
print(results_df)
```
